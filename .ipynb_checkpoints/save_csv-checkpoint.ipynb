## Salvando o arquivo PCA.csv com arquivo separado por vírgulas

```python
nome_arquivo = 'PCA.csv'    #Substituir pelo nome do arquivo
df = pd.read_csv(f'./dados_pca/{nome_arquivo}', sep=';', encoding='iso8859')
# df.head()
```

```python
df.to_csv("./dados_pca/PCA_virgula.csv", sep=",", index=False, encoding="utf-8")
# Apague o PCA.csv original, que estava com ';', e renomeie esse arquivo PCA_virgula.csv para PCA.csv
```
